```python
# ============================================================
# 10. LOAD TRAINED ML MODEL
# ============================================================

import joblib

MODEL_PATH = r"C:\Users\ANTARA\OneDrive\Documents\Career\Dynamic Pricing Intelligence System\best_pricing_model.pkl"

FEATURES_PATH = r"C:\Users\ANTARA\OneDrive\Documents\Career\Dynamic Pricing Intelligence System\model_features.pkl"

# Load trained Random Forest model
pricing_model = joblib.load(MODEL_PATH)

# Load feature order
model_features = joblib.load(FEATURES_PATH)

print("✅ ML model loaded successfully!")
print("Model:", type(pricing_model).__name__)
print("Number of features:", len(model_features))
print("Features:", model_features)
```

```python
# ============================================================
# 11. CHATBOT - ML SALES PREDICTION
# ============================================================

import re

def predict_store_sales(store_number):

    # Get data for the selected store
    store_data = pricing_df[
        pricing_df["Store"] == store_number
    ]

    if len(store_data) == 0:
        return None

    # Take one existing row from the store
    input_data = store_data.iloc[[0]].copy()

    # Remove target and date
    input_data = input_data.drop(
        columns=["Sales", "Date"]
    )

    # Ensure exact feature order
    input_data = input_data[model_features]

    # Predict sales
    prediction = pricing_model.predict(input_data)[0]

    return prediction


print("✅ ML prediction function created successfully!")
```

```python
# ============================================================
# 13. DYNAMIC PRICING - SCENARIO SIMULATION
# ============================================================

def pricing_scenario(store_number, current_price):
    """
    Compare simulated pricing scenarios for a store.

    Note:
    The current dataset does not contain an actual price column.
    Therefore, price impact is simulated rather than learned
    directly from historical price data.
    """

    # Find store data
    store_data = pricing_df[
        pricing_df["Store"] == store_number
    ]

    if len(store_data) == 0:
        return None

    # Use an existing row as the store's current situation
    base_data = store_data.iloc[[0]].copy()

    # Remove target and date
    base_data = base_data.drop(
        columns=["Sales", "Date"]
    )

    # Keep exact model feature order
    base_data = base_data[model_features]

    # Base demand prediction
    base_prediction = pricing_model.predict(base_data)[0]

    # Simulated price scenarios
    price_scenarios = {
        "10% Lower": current_price * 0.90,
        "Current": current_price,
        "10% Higher": current_price * 1.10
    }

    results = []

    for scenario, price in price_scenarios.items():

        # Simple demand sensitivity assumption
        if scenario == "10% Lower":
            demand_factor = 1.05

        elif scenario == "10% Higher":
            demand_factor = 0.95

        else:
            demand_factor = 1.00

        simulated_demand = base_prediction * demand_factor

        simulated_revenue = price * simulated_demand

        results.append({
            "Scenario": scenario,
            "Price": round(price, 2),
            "Predicted Demand": round(simulated_demand, 2),
            "Simulated Revenue": round(simulated_revenue, 2)
        })

    return pd.DataFrame(results)


print("✅ Dynamic pricing scenario function created!")
```

```python
scenario_result = pricing_scenario(
    store_number=10,
    current_price=100
)

display(scenario_result)
```

```python
# ============================================================
# 14. DYNAMIC PRICING - BEST PRICE RECOMMENDATION
# ============================================================

def recommend_price(store_number, current_price):
    """
    Recommend the price scenario with the highest
    simulated revenue.
    """

    scenario_result = pricing_scenario(
        store_number,
        current_price
    )

    if scenario_result is None:
        return None

    # Find scenario with maximum simulated revenue
    best_row = scenario_result.loc[
        scenario_result["Simulated Revenue"].idxmax()
    ]

    print(f"📍 Store: {store_number}")
    print(f"💰 Current Price: ₹{current_price:,.2f}")
    print(
        f"⭐ Recommended Scenario: "
        f"{best_row['Scenario']}"
    )
    print(
        f"💵 Recommended Price: "
        f"₹{best_row['Price']:,.2f}"
    )
    print(
        f"📈 Predicted Demand: "
        f"{best_row['Predicted Demand']:,.2f}"
    )
    print(
        f"💰 Simulated Revenue: "
        f"₹{best_row['Simulated Revenue']:,.2f}"
    )

    return best_row


print("✅ Price recommendation function created!")
```

```python
# ============================================================
# 15. CHATBOT - PRICING RECOMMENDATION
# ============================================================

print("🤖 Dynamic Pricing Intelligence Chatbot")
print("Ask me for a price recommendation.")
print("Example: Recommend a price for Store 10 at ₹100")
print("Type 'exit' to stop.\n")

while True:

    user_input = input("You: ").strip()
    question = user_input.lower()

    if question == "exit":
        print("Bot: Goodbye! 👋")
        break

    # --------------------------------------------------------
    # PRICE RECOMMENDATION
    # --------------------------------------------------------
    if (
        "recommend a price" in question
        or "price recommendation" in question
        or "best price" in question
        or "recommend price" in question
    ):

        # Extract store number
        store_match = re.search(
            r"store\s*(\d+)",
            question,
            re.IGNORECASE
        )

        # Extract price
        price_match = re.search(
            r"(?:₹|rs\.?|inr)?\s*(\d+(?:\.\d+)?)",
            question,
            re.IGNORECASE
        )

        if store_match and price_match:

            store_number = int(store_match.group(1))
            current_price = float(price_match.group(1))

            recommendation = recommend_price(
                store_number,
                current_price
            )

            if recommendation is None:

                print(
                    f"Bot: I couldn't find Store "
                    f"{store_number} in the dataset."
                )

        elif not store_match:

            print(
                "Bot: Please mention a store number."
            )

        elif not price_match:

            print(
                "Bot: Please provide the current price."
            )

    # --------------------------------------------------------
    # UNKNOWN QUESTION
    # --------------------------------------------------------
    else:

        print(
            "Bot: I can currently provide price "
            "recommendations for a store."
        )
```

```python
# ========================================================
# PRICING SCENARIO FUNCTION
# ========================================================

def pricing_scenario(store_number, current_price):

    store_data = pricing_df[
        pricing_df["Store"] == store_number
    ]

    if len(store_data) == 0:
        return None

    base_data = store_data.iloc[[0]].copy()

    base_data = base_data.drop(
        columns=["Sales", "Date"]
    )

    base_data = base_data[model_features]

    base_prediction = pricing_model.predict(
        base_data
    )[0]

    price_scenarios = {
        "10% Lower": current_price * 0.90,
        "Current": current_price,
        "10% Higher": current_price * 1.10
    }

    results = []

    for scenario, price in price_scenarios.items():

        if scenario == "10% Lower":
            demand_factor = 1.05

        elif scenario == "10% Higher":
            demand_factor = 0.95

        else:
            demand_factor = 1.00

        simulated_demand = (
            base_prediction * demand_factor
        )

        simulated_revenue = (
            price * simulated_demand
        )

        results.append({
            "Scenario": scenario,
            "Price": round(price, 2),
            "Predicted Demand": round(simulated_demand, 2),
            "Simulated Revenue": round(simulated_revenue, 2)
        })

    return pd.DataFrame(results)


print("✅ Pricing scenario function created!")
```

```python
# ========================================================
# STEP 30 - PROMOTION BUSINESS RECOMMENDATION
# ========================================================

def promotion_recommendation():

    result = get_promotion_insight()

    if result is None:
        return None

    promo_sales = result["promo_sales"]
    non_promo_sales = result["non_promo_sales"]
    percentage_difference = result["percentage_difference"]

    if promo_sales > non_promo_sales:

        recommendation = (
            f"Promotions appear effective. "
            f"Average sales during promotion days are "
            f"{percentage_difference:.2f}% higher than "
            f"non-promotion days."
        )

    else:

        recommendation = (
            f"Promotions do not appear to improve sales. "
            f"Average sales are "
            f"{abs(percentage_difference):.2f}% lower "
            f"during promotion days."
        )

    return recommendation


print("✅ Promotion recommendation function created!")
```

```python
# ========================================================
# PROMOTION INSIGHT FUNCTION
# ========================================================

def get_promotion_insight():

    promo_sales = pricing_df[
        pricing_df["Promo"] == 1
    ]["Sales"].mean()

    non_promo_sales = pricing_df[
        pricing_df["Promo"] == 0
    ]["Sales"].mean()

    difference = promo_sales - non_promo_sales

    percentage_difference = (
        difference / non_promo_sales
    ) * 100

    if difference > 0:

        insight = (
            f"Sales are {percentage_difference:.2f}% higher "
            "on average during promotion days."
        )

    else:

        insight = (
            f"Sales are {abs(percentage_difference):.2f}% lower "
            "on average during promotion days."
        )

    return {
        "promo_sales": promo_sales,
        "non_promo_sales": non_promo_sales,
        "difference": difference,
        "percentage_difference": percentage_difference,
        "insight": insight
    }


print("✅ Promotion insight function loaded!")
```
